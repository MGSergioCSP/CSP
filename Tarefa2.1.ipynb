### Exercicio 1: Fundamentos de Clase e Obxectos

- Define unha clase chamada "Rectangulo" cun construtor que reciba os parámetros: "lonxitude" e "ancho". Gárdaos en atributos de instancia.
- Engade un método chamado "area" que calcule e devolva a área do rectángulo.
- Crea dous obxectos (instancias) de "Rectangulo" con diferentes medidas, calcula a súa área e imprime o seu diccionario interno de atributos.


```python
class Rectangulo:
    def __init__(self, lonxitude, ancho):
        self.lonxitude = lonxitude
        self.ancho = ancho
        
    def area(self):
        return self.lonxitude * self.ancho

rectangulo1 = Rectangulo(5, 3)
rectangulo2 = Rectangulo(10.5, 4)

print(f"Área de rectángulo 1: {rectangulo1.area()}")
print(f"Área de rectángulo 2: {rectangulo2.area()}")
print(f"Dicionario interno de rectángulo 1: {rectangulo1.__dict__}")
print(f"Dicionario interno de rectángulo 2: {rectangulo2.__dict__}")
```

### Exercicio 2: Atributos "Privados" e Propiedades

- Crea unha clase "ContaBancaria".
- No constructor (`__init__`), inicializa un atributo "privado" denominado "saldo". Que non se poida acceder directamente a el.
- Crea un "getter" para poder consultar dito saldo.
- Crea un "setter" que, ao intentar modificar o saldo, só permita asignar valores positivos e "racionales".
- Fai probas de uso.

```python
class ContaBancaria:
    def __init__(self, saldo_inicial=0):

        self.__saldo = 0
        self.saldo = saldo_inicial 

    @property
    def saldo(self):
        """Devolve o saldo actual da conta"""
        return self.__saldo

    @saldo.setter
    def saldo(self, valor):
        """Modifica o saldo, só se é un valor numérico (racional/real) e positivo"""

        if isinstance(valor, (int, float)) and valor >= 0:
            self.__saldo = valor
        else:
            print(f"Erro: Ingreso rexeitado. O valor '{valor}' debe ser un número positivo.")

print("--- Creando a conta ---")
conta = ContaBancaria(100.50)
print(f"Saldo inicial correcto: {conta.saldo}")

print("\n--- Modificando o saldo correctamente ---")
conta.saldo = 500
print(f"Novo saldo: {conta.saldo}")

print("\n--- Intentando asignar un valor negativo ---")
conta.saldo = -50
print(f"O saldo mantense en: {conta.saldo}")

print("\n--- Intentando asignar un valor non numérico (texto) ---")
conta.saldo = "trescentos"
print(f"O saldo mantense en: {conta.saldo}")

print("\n--- Intentando acceder directamente ao atributo privado ---")
try:
    print(conta.__saldo)
except AttributeError as e:
    print(f"Erro interceptado: Non se pode acceder a __saldo directamente. ({e})")
```

### Exercicio 3: Herdanza Simple e Especialización

- A partires da clase "Pila" proporcionada nos contidos que ten un atributo privado "__stack_list" e métodos "push" e "pop",  crea unha nova clase denominada "PilaSegura" que herde de "Pila".

- Utiliza métodos coma "issubclass" e "isinstance" para comprobar a relación.


```python
class Pila():
    def __init__(self): 
        self.__description = "A miña pila de proba" 
        self.__stack_list = [] 

    @property
    def description(self):  
        return self.__description

    @description.setter  
    def description(self, descr):
        self.__description = descr

    def push(self, val):
        self.__stack_list.append(val)

    def pop(self):
        if self.__stack_list == []:
            raise IndexError("Pop en lista baleira")
        else:
            val = self.__stack_list[-1]
            del self.__stack_list[-1]
            return val

class PilaSegura(Pila):
    
    def pop(self):
        val = super().pop()
        print("Elemento retirado correctamente.")
        return val
        
    def esta_baleira(self):
        return len(self._Pila__stack_list) == 0


print("--- Creando e enchendo a PilaSegura ---")
miña_pila_segura = PilaSegura()

print(f"¿A pila está baleira ao comezo? -> {miña_pila_segura.esta_baleira()}")

miña_pila_segura.push("Documento 1")
miña_pila_segura.push("Documento 2")

print(f"¿A pila está baleira despois de engadir elementos? -> {miña_pila_segura.esta_baleira()}")

print("\n--- Retirando elementos ---")
item1 = miña_pila_segura.pop() 
print(f"O elemento que sacamos é: '{item1}'")

item2 = miña_pila_segura.pop()
print(f"O elemento que sacamos é: '{item2}'")

print(f"\n¿E agora, a pila está baleira? -> {miña_pila_segura.esta_baleira()}")

print("\n--- Comprobando as relacións de Herdanza ---")

print(f"¿'PilaSegura' é subclase de 'Pila'? -> {issubclass(PilaSegura, Pila)}")
print(f"¿'Pila' é subclase de 'PilaSegura'? -> {issubclass(Pila, PilaSegura)}")

print(f"¿O obxecto 'miña_pila_segura' é unha instancia de PilaSegura? -> {isinstance(miña_pila_segura, PilaSegura)}")
print(f"¿O obxecto 'miña_pila_segura' é unha instancia de Pila (por herdanza)? -> {isinstance(miña_pila_segura, Pila)}")
```

### Exercicio 4: Atributos de Clase vs. Instancia

- Crea unha clase "Coche" cun atributo de clase denominado "numero_de_rodas" e con valor = 4.
- No constructor define un atributo de instancia denominado "marca" que se reciba coma parámetro.
- Ten tamén un atributo de clase denominado "total_coches_creados" que leve a conta de todas as instancias xeradas.
- Crea 5 obxectos "Coche" de distintas marcas.
- Imprime, en cada caso, os atributos de clase e o diccionario de cada instancia.


```python
class Coche:
    numero_de_rodas = 4
    total_coches_creados = 0
    
    def __init__(self, marca):
        self.marca = marca
  
        Coche.total_coches_creados += 1


marcas_a_crear = ["Seat", "Renault", "Ford", "Toyota", "Volkswagen"]
lista_de_coches = []

print("--- Creando coches e revisando os seus datos ---")

for marca in marcas_a_crear:
    novo_coche = Coche(marca)
    lista_de_coches.append(novo_coche)

    print(f"\nAcabase de crear un coche marca: {novo_coche.marca}")

    print(f"- Atributo de clase (número de rodas): {Coche.numero_de_rodas}")
    print(f"- Atributo de clase (total creados agora mesmo): {Coche.total_coches_creados}")

    print(f"- Diccionario interno da instancia: {novo_coche.__dict__}")


print("\n--- Resumo final ---")
print(f"Total de coches creados rexistrados na clase: {Coche.total_coches_creados}")
```

### Exercicio 5: Métodos Estáticos, de Clase e de Instancia

- Crea unha clase denominada "UtilidadesMatematicas" cos seguintes métodos:
    - Un método estático denominado "e_par" que devolva True ou False.
    - Un método de clase denominado "informacion" que devolva unha cadea descritiva sobre a clase.
    - Un método de instancia denominado "potencia" que, para un número base almacenado nun atributo de instancia inicializado no constructor, calcule a potencia a dito exponente.
- Fai o preciso para chamar a todos os métodos creados.

```python
class UtilidadesMatematicas:
    
    def __init__(self, base):
        self.base = base
    @staticmethod
    def e_par(numero):
        """Devolve True se o número é par, e False en caso contrario."""
        return numero % 2 == 0

    @classmethod
    def informacion(cls):
        """Devolve unha cadea descritiva da clase."""
        return f"Esta é a clase '{cls.__name__}'. Contén funcións de utilidade matemática."

    def potencia(self, exponente):
        """Calcula a potencia da base (gardada na instancia) elevada ao expoñente dado."""
        return self.base ** exponente

print("--- 1. Probando o Método Estático ---")
numero_proba = 7
print(f"¿O número {numero_proba} é par? -> {UtilidadesMatematicas.e_par(numero_proba)}")
print(f"¿O número 8 é par? -> {UtilidadesMatematicas.e_par(8)}")

print("\n--- 2. Probando o Método de Clase ---")
print(f"Información: {UtilidadesMatematicas.informacion()}")

print("\n--- 3. Probando o Método de Instancia ---")
calculadora = UtilidadesMatematicas(base=5) 
exponente_proba = 3

resultado_potencia = calculadora.potencia(exponente_proba)

print(f"O obxecto 'calculadora' ten a base: {calculadora.base}")
print(f"O resultado de {calculadora.base} elevado a {exponente_proba} é: {resultado_potencia}")

calculadora.base = 2
print(f"Nova base {calculadora.base} elevada a 10 é: {calculadora.potencia(10)}")
```

### Exercicio 6: Dunder Methods

- Crea unha clase "Vector2D" para representar un vector no plano cun constructor que reciba as coordenadas "x" e "y".
- Implementa (ou sobrescribe) o método "__add__" para que poidas sumar dous vectores usando o operador "+". Deberá devolver un obxecto novo coa suma das coordenadas.
- Implementa (ou sobrescribe) o método "__str__" para que, cando se convirta o vector a string, amose un formato lexible como "Vector2D(x=5, y=3)".
- Implementa (ou sobrescribe) o método "__eq__" para que dous vectores sexan iguais se as súas coordenadas "x" e "y" coinciden.
- Constrúe exemplos de uso.

```python
class Vector2D:

    def __init__(self, x, y):
        self.x = x
        self.y = y


    def __add__(self, outro_vector):

        if isinstance(outro_vector, Vector2D):
            novo_x = self.x + outro_vector.x
            novo_y = self.y + outro_vector.y
            return Vector2D(novo_x, novo_y)
        else:
            raise TypeError("Só se poden sumar dous obxectos de tipo Vector2D")

    def __str__(self):
        return f"Vector2D(x={self.x}, y={self.y})"

    def __eq__(self, outro_vector):
        if isinstance(outro_vector, Vector2D):
            return self.x == outro_vector.x and self.y == outro_vector.y
        return False



print("--- Creando os vectores ---")
v1 = Vector2D(2, 4)
v2 = Vector2D(3, 1)
v3 = Vector2D(2, 4) 

print(f"Vector 1: {v1}")
print(f"Vector 2: {v2}")
print(f"Vector 3: {v3}")

print("\n--- Probando a suma (+) ---")
v_suma = v1 + v2
print(f"A suma de {v1} + {v2} é igual a: \n -> {v_suma}")

print("\n--- Probando a igualdade (==) ---")
print(f"¿O Vector 1 é igual ao Vector 2? -> {v1 == v2}")
print(f"¿O Vector 1 é igual ao Vector 3? -> {v1 == v3}")

print("\n--- Outra proba da suma ---")
v_total = v1 + v2 + v3 
print(f"A suma (v1 + v2 + v3) resulta en: {v_total}")
```

### Exercicio 7: Herdanza Múltiple e Dunder Methods

- Define unha clase simple "Imprimible" co método "amosar" que devolva "Son un obxecto imprimible".
- A partires da clase anterior "Pila", haberá outra que se chame "PilaSuma" que herde de "Pila". crea unha nova clase "PilaSumaImprimible" que herde de ambas: "PilaSuma" e "Imprimible".
- No constructor de "PilaSumaImprimible" inicializa correctamente a parte correspondente de "PilaSuma".
- Sobrescribe o método "amosar" para que, no lugar do texto xenérico, devolva un string informativo como "PilaSuma coa suma total: X", onde "X" será o valor da propiedad "suma".
- Crea unha instancia desta clase, engade algúns números co "push", e proba a chamar tanto aos métodos herdados ("pop" e "suma") como ao método personalizado "amosar".
- Que fai "__mro__"? E "__bases__"? Utiliza métodos como issubclass.

```python

class Pila:
    def __init__(self):
        self.__stack_list = []
        
    def push(self, val):
        self.__stack_list.append(val)
        
    def pop(self):
        if self.__stack_list == []:
            raise IndexError("Pop en lista baleira")
        else:
            val = self.__stack_list[-1]
            del self.__stack_list[-1]
            return val

class PilaSuma(Pila):
    def __init__(self):
        super().__init__()
        self.__suma = 0
        
    def push(self, val):
        super().push(val)
        self.__suma += val
        
    def pop(self):
        val = super().pop()
        self.__suma -= val
        return val
        
    @property
    def suma(self):
        return self.__suma

class Imprimible:
    def amosar(self):
        return "Son un obxecto imprimible"


class PilaSumaImprimible(PilaSuma, Imprimible):
    
    def __init__(self):
        super().__init__()

    def amosar(self):
        return f"PilaSuma coa suma total: {self.suma}"


print("--- 1. Creando a Pila e usando Push ---")
miña_pila_mixta = PilaSumaImprimible()
miña_pila_mixta.push(10)
miña_pila_mixta.push(20)
miña_pila_mixta.push(5)
print("Engadeuse o 10, o 20 e o 5.")

print("\n--- 2. Probando os métodos herdados e o amosar personalizado ---")
print(f"Suma actual (herdada): {miña_pila_mixta.suma}")

print(f"Mensaxe do método amosar: '{miña_pila_mixta.amosar()}'")

print("\n--- 3. Probando Pop ---")
valor_sacado = miña_pila_mixta.pop()
print(f"Sacamos un valor: {valor_sacado}")
print(f"Nova mensaxe do método amosar: '{miña_pila_mixta.amosar()}'")

print("\n--- 4. Introspección (MRO, bases, issubclass) ---")

print(f"¿PilaSumaImprimible herda de PilaSuma? -> {issubclass(PilaSumaImprimible, PilaSuma)}")
print(f"¿PilaSumaImprimible herda de Imprimible? -> {issubclass(PilaSumaImprimible, Imprimible)}")
print(f"¿PilaSumaImprimible herda de Pila (O Avó)? -> {issubclass(PilaSumaImprimible, Pila)}")

print("\n__bases__ (Nais directas):")
print(PilaSumaImprimible.__bases__)

print("\n__mro__ (Camino completo de busca de métodos e herdanza):")
for clase in PilaSumaImprimible.__mro__:
    print(f"  -> {clase.__name__}")
```
