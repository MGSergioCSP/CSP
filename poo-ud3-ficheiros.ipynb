### Apertura Segura con try...except:

* Escribe unha función chamada `abrir_ficheiro_seguro(nome_ficheiro, modo)` que intente abrir un ficheiro no modo especificado.
* Debe usar un bloque `try...except` para capturar calquera excepción (como `FileNotFoundError`).
* Se a apertura é exitosa, a función debe devolver o obxecto do ficheiro. Se hai un erro, debe amosar a mensaxe de erro e devolver `None`.
* Fai probas invocando a función cun ficheiro que non exista e despois cun que si exista (por exemplo, en modo lectura `"r"`). Non esquezas pechar o ficheiro se se abriu correctamente.

````python
def abrir_ficheiro_seguro(nome_ficheiro, modo):
    try:
        ficheiro = open(nome_ficheiro, modo)
        return ficheiro
    except FileNotFoundError:
        print(f"Erro: O ficheiro '{nome_ficheiro}' non existe.")
        return None
    except Exception as e:
        print(f"Erro inesperado: {e}")
        return None

ficheiro_falso = abrir_ficheiro_seguro("ficheiro_que_non_existe.txt", "r")

with open("ficheiro_real.txt", "w") as f:
    f.write("Ola mundo")
    
ficheiro_real = abrir_ficheiro_seguro("ficheiro_real.txt", "r")
if ficheiro_real:
    print("Ficheiro aberto con éxito.")
    ficheiro_real.close()
````

### Diario de Notas Persoal

* Crea un programa que permita ao usuario escribir anotacións nun ficheiro chamado `diario.txt`.
* O programa debe:
    1. Preguntar ao usuario se quere (1) Ler todas as anotacións ou (2) Engadir unha nova anotación.
    2. Se escolle ler, debe abrir o ficheiro en modo lectura (`"r"`) e amosar todo o seu contido na pantalla. Controla o erro se o ficheiro aínda non existe.
    3. Se escolle engadir, debe pedirlle ao usuario unha liña de texto e **engadila** ao final do ficheiro (modo `"a"`), incluíndo a data e hora actual ao comezo da liña (usa o módulo `datetime`).


````python
import datetime
import os

def diario_notas():
    ficheiro = "diario.txt"
    opcion = input("Queres (1) Ler todas as anotacións ou (2) Engadir unha nova anotación? (1/2): ")
    
    if opcion == "1":
        if os.path.exists(ficheiro):
            with open(ficheiro, "r") as f:
                contido = f.read()
                print("\n--- Contido do diario ---")
                print(contido if contido else "O diario está baldeiro.")
        else:
            print("O ficheiro aínda non existe.")
            
    elif opcion == "2":
        anotacion = input("Escribe a túa anotación: ")
        data_hora_actual = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        with open(ficheiro, "a") as f:
            f.write(f"[{data_hora_actual}] {anotacion}\n")
        print("Anotación gardada con éxito.")
        
    else:
        print("Opción non válida.")

diario_notas()
````

### Analizador de Ficheiro de Texto

* Pídelle ao usuario o nome dun ficheiro de texto.
* O programa debe abrir o ficheiro e calcular:
    * Número total de liñas.
    * Número total de palabras.
    * Número total de caracteres.
* Crea un ficheiro `proba.txt` con algunhas liñas de texto e comproba que os cálculos son correctos.

````python
def analizador_ficheiro():
    with open("proba.txt", "w") as f:
        f.write("Esta e a primeira lina.\nAqui esta a segunda lina.\nE unha terceira.")

    nome_ficheiro = input("Introduce o nome do ficheiro de texto a analizar (ex. proba.txt): ")
    try:
        with open(nome_ficheiro, "r", encoding="utf-8") as f:
            linas = f.readlines()
            
        num_linas = len(linas)
        num_palabras = sum(len(lina.split()) for lina in linas) 
        num_caracteres = sum(len(lina) for lina in linas) 
        
        print(f"Liñas totais: {num_linas}")
        print(f"Palabras totais: {num_palabras}")
        print(f"Caracteres totais: {num_caracteres}")
        
    except FileNotFoundError:
         print("Erro: O ficheiro especificado non foi atopado.")

analizador_ficheiro()
````

### Copiador de Ficheiros con Número de Liña

* Pídelle ao usuario o nome dun ficheiro de orixe.
* Crea un novo ficheiro denominado `copia_con_numeros.txt`.
* Le o ficheiro de orixe **liña por liña** e escribe cada liña no novo ficheiro, pero **engadíndolle o número de liña ao principio**.
* Por exemplo, se unha liña orixinal é "Hola mundo", no novo ficheiro debe aparecer "1 Hola mundo".

````python
def copiar_con_numeros():
    orixe = input("Introduce o nome do ficheiro de orixe: ")
    destino = "copia_con_numeros.txt"
    
    try:
        with open(orixe, "r", encoding="utf-8") as f_orixe, open(destino, "w", encoding="utf-8") as f_destino:
            for numero, lina in enumerate(f_orixe, start=1):
                f_destino.write(f"{numero} {lina}")
                
        print(f"Ficheiro copiado con éxito nnovo ficheiro '{destino}'.")
    except FileNotFoundError:
        print("Erro: O ficheiro de orixe non existe.")

copiar_con_numeros()
````

### Estracto bancario en CSV

* Crea un programa para ler e almacenar un estracto bancario do teu banco en CSV.
* Os pasos que terás que dar son os seguintes:
    * Descarga un ficheiro CSV do teu banco.
    * Instala `chardet` para detectar o encoding do teu ficheiro.
    * Averigua as cabeceiras. Por exemplo: `Data ctble;Data valor;Concepto;Importe;Moeda;Saldo;Moeda;Concepto ampliado`. Un exemplo sería: `02-01-2026;29-12-2025;123456789123 PEIXERIA TRINQUETE 123456789;-14,97;EUR;1234,62;EUR;`
    * O programa debe ter un menú para:
        * Cargar os datos dun CSV nun Diccionario ou, mellor dito, nunha lista de Diccionarios.
        * Buscar por concepto e amosar liña a liña todos os conceptos que conteñan a cadea introducida.
        * Buscar por cantidade e amosar liña a liña todas as cantidades que coincidan.
        * Suma por conceptos. Se lle introduce un concepto e amosa a suma de todas as entradas que conteñan ese concepto.
        * Procurar movementos entre datas e amosalos por pantalla.
    * O programa só pode rematar cando se lle indique a través dunha opción de menú.

````python
import csv
from datetime import datetime


def xestor_bancario_csv():
    ficheiro = "estracto.csv"
    movementos = []
    
    try:
        with open(ficheiro, mode="r", encoding="utf-8") as f: 
            lector = csv.DictReader(f, delimiter=";")
            for fila in lector:
                movementos.append(fila)
    except FileNotFoundError:
        print("Ficheiro CSV non atopado. Crea un estracto.csv primeiro para probar.")
        return
        
    while True:
        print("\n--- Menú Estracto Bancario ---")
        print("1. Buscar por concepto")
        print("2. Buscar por cantidade")
        print("3. Suma por concepto")
        print("4. Procurar movementos entre datas")
        print("5. Saír")
        opcion = input("Elixe unha opción: ")
        
        if opcion == "1":
            concepto_busca = input("Introduce o concepto a buscar: ").lower()
            atopados = [m for m in movementos if conceito_busca in m.get("Concepto", "").lower()]
            for m in atopados: print(m)
                
        elif opcion == "2":
            cantidade_busca = input("Introduce a cantidade a buscar: ")
            atopados = [m for m in movementos if m.get("Importe") == cantidade_busca]
            for m in atopados: print(m)
                
        elif opcion == "3":
            concepto_busca = input("Introduce o concepto a sumar: ").lower()
            suma = 0.0
            for m in movementos:
                if concepto_busca in m.get("Concepto", "").lower():
                    importe_str = m.get("Importe", "0").replace(",", ".")   
                    try:
                        suma += float(importe_str)
                    except ValueError:
                        pass
            print(f"Suma total para '{concepto_busca}': {suma:.2f}")
            
        elif opcion == "4":
            data_inicio_str = input("Data inicio (DD-MM-YYYY): ")
            data_fin_str = input("Data fin (DD-MM-YYYY): ")
            try:
                data_inicio = datetime.strptime(data_inicio_str, "%d-%m-%Y")
                data_fin = datetime.strptime(data_fin_str, "%d-%m-%Y")
                
                for m in movementos:
                    data_movimento = datetime.strptime(m.get("Data ctble", "01-01-1970"), "%d-%m-%Y")
                    if data_inicio <= data_movimento <= data_fin:
                        print(m)
            except ValueError:
                print("Formato de data incorrecto.")
                
        elif opcion == "5":
            break
        else:
            print("Opción incorrecta.")

xestor_bancario_csv()
````

###  Xestor de Contactos en JSON

* Crea un programa para xestionar unha axenda de contactos nun ficheiro de texto `contactos.JSON`.
* A estrutura será unha lista de dicionarios, onde cada dicionario ten as chaves `nome`, `telefono`, `email`.
* O programa debe ter un menú para:
    * **Engadir contacto**: Pide os datos e engádeos ao ficheiro.
    * **Buscar contacto**: Pide un nome e amosa o teléfono e email de todas as liñas que conteñan ese nome.
    * **Listar todos**: Amosa todos os contactos numerados.

````python
import json
import os

def xestor_contactos():
    ficheiro = "contactos.json"
    
    if os.path.exists(ficheiro):
        with open(ficheiro, "r") as f:
            try:
                contactos = json.load(f)
            except json.JSONDecodeError:
                contactos = []
    else:
        contactos = []

    while True:
        print("\n--- Axenda de Contactos ---")
        print("1. Engadir contacto")
        print("2. Buscar contacto")
        print("3. Listar todos")
        print("4. Saír")
        opcion = input("Opción: ")
        
        if opcion == "1":
            nome = input("Nome: ")
            telefono = input("Teléfono: ")
            email = input("Email: ")
            
            contactos.append({"nome": nome, "telefono": telefono, "email": email})
            
            with open(ficheiro, "w") as f:
                json.dump(contactos, f, indent=4)
            print("Contacto gardado exitosamente.")
            
        elif opcion == "2":
            busca = input("Dime o nome a buscar: ").lower()
            atopados = [c for c in contactos if busca in c["nome"].lower()]
            
            if not atopados:
                print("Non se atoparon resultados.")
            for c in atopados:
                print(f"Teléfono: {c['telefono']} | Email: {c['email']}")
                
        elif opcion == "3":
            for i, c in enumerate(contactos, 1):
                print(f"{i}. Nome: {c['nome']}  |  Tel: {c['telefono']}  |  Email: {c['email']}")
                
        elif opcion == "4":
            print("Saindo...")
            break
        else:
            print("Opción incorrecta.")

xestor_contactos()
````

### Cálculo con Datos Numéricos usando NumPy

* Crea un ficheiro de texto chamado `matriz.txt` con varias liñas de números separados por espazos. Por exemplo:
    ```
    1.5 2.3 4.8
    5.1 3.2 7.4
    9.6 8.7 6.5
    ```
* Escribe un script que:
    1. Cargue os datos do ficheiro nun array de **NumPy**.
    2. Calcule e amose por pantalla: a media de todos os números, a suma de cada fila e o valor máximo de cada columna. 
    3. Finalmente, garde o array orixinal multiplicado por 2 nun novo ficheiro `matriz_dobre.txt` con formato `'%.2f'`.

````python
import numpy as np

def calculo_numpy():
    texto_matriz = "1.5 2.3 4.8\n5.1 3.2 7.4\n9.6 8.7 6.5"
    with open("matriz.txt", "w") as f:
        f.write(texto_matriz)
        
    try:
        datos = np.loadtxt("matriz.txt")
        print("Matriz orixinal (cargada):")
        print(datos)
        
        print(f"\nMedia de todos os números: {np.mean(datos):.2f}")
        print(f"Suma de cada fila (axis=1): {np.sum(datos, axis=1)}")
        print(f"Valor máximo de cada columna (axis=0): {np.max(datos, axis=0)}")
        
        datos_dobre = datos * 2
        np.savetxt("matriz_dobre.txt", datos_dobre, fmt='%.2f')
        print(f"\nMatriz gardada en 'matriz_dobre.txt' con éxito.")
        
    except Exception as e:
        print(f"Atopouse un erro traballando con NumPy: {e}")

calculo_numpy()
````
